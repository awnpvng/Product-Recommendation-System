```python
import polars as pl
import polars.selectors as cs
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency
```

```python
data_path = '/kaggle/input/dataset'
```

```python
item = pl.read_parquet(data_path + '/sales_pers.item_chunk_0.parquet')
item
```

```python
purchase_history = pl.read_parquet(data_path + '/sales_pers.purchase_history_daily_chunk_*.parquet')
purchase_history
```

```python
user = pl.read_parquet(data_path + '/sales_pers.user_chunk_*.parquet')
user
```

```python
if "date_key" not in purchase_history.columns:
    purchase_history = purchase_history.with_columns(
        pl.col("timestamp").str.strptime(pl.Datetime, strict=False).dt.date().alias("date_key")
    )

b = purchase_history.select(["customer_id", "date_key", "item_id"]).drop_nulls().unique()

pairs = (
    b.join(b, on=["customer_id", "date_key"], suffix="_r")
     .filter(pl.col("item_id") < pl.col("item_id_r"))          # bỏ trùng và cặp đảo
     .group_by(["item_id", "item_id_r"])
     .agg(pl.len().alias("#cooc"))
     .rename({"item_id": "item_1", "item_id_r": "item_2"})
     .sort("#cooc", descending=True)
)

pairs
```

```python
#lọc thông tin và age_group
item.select(pl.col("age_group").unique()).to_series().to_list()
```

```python


# Gom text để dò "step 1" / "mom" và chuẩn hoá age_group
txt_cols = ["description", "brand", "category", "category_l1", "category_l2", "category_l3", "item_type", "description_new"]

item_signals = (
    item.with_columns([
        pl.concat_str(
            [pl.col(c).cast(pl.Utf8) for c in txt_cols if c in item.columns],
            separator=" ",
            ignore_nulls=True
        ).str.to_lowercase().alias("full_text"),
        pl.col("age_group").cast(pl.Utf8).str.to_lowercase().alias("age_group_norm"),
    ])
    .select(["item_id", "full_text", "age_group_norm"])
)
```

```python
# Chỉ giữ các cột cần và đảm bảo timestamp là i64
ph_base = purchase_history.select(["customer_id", "item_id", "timestamp"])
# Nếu cần chắc chắn i64:
# ph_base = ph_base.with_columns(pl.col("timestamp").cast(pl.Int64))
```

```python
# Join nhẹ: chỉ 3 cột purchase + 3 cột tín hiệu từ item
phj = ph_base.join(item_signals, on="item_id", how="left")

# Điều kiện tín hiệu (đã lowercase)
is_step1 = pl.col("full_text").str.contains(r"\bstep[\s-]?1\b")
is_mom   = pl.col("full_text").str.contains(r"\bmom\b")
# Nhiều biến thể 0–3M
is_age03 = pl.col("age_group_norm").str.contains(r"0\s*[-–]\s*3\s*m|0-3m|0-3 m")

by_user_dates = phj.group_by("customer_id").agg([
    pl.when(is_step1).then(pl.col("timestamp")).min().alias("first_date_buy_step1"),
    pl.when(is_age03).then(pl.col("timestamp")).min().alias("first_date_buy_age_group_0-3M"),
    pl.when(is_mom).then(pl.col("timestamp")).max().alias("last_date_buy_milk4mom"),
])

by_user_dates
```

```python
pred_table = (
    by_user_dates.with_columns([
        pl.when(pl.col("first_date_buy_step1").is_not_null())
          .then(pl.lit("0-6M (from 'Step 1')")).otherwise(None).alias("age_by_step1"),
        pl.when(pl.col("first_date_buy_age_group_0-3M").is_not_null())
          .then(pl.lit("0-3M (from age_group)")).otherwise(None).alias("age_by_age_group"),
        pl.when(pl.col("last_date_buy_milk4mom").is_not_null())
          .then(pl.lit("prenatal/postpartum (from 'Mom')")).otherwise(None).alias("age_by_milk4mom"),
    ])
    .select([
        "customer_id",
        "first_date_buy_step1", "age_by_step1",
        "first_date_buy_age_group_0-3M", "age_by_age_group",
        "last_date_buy_milk4mom", "age_by_milk4mom"
    ])
    # 🔽 Chỉ giữ user có ít nhất 1 trong 3 cột thời gian khác null
    .filter(
        pl.col("first_date_buy_step1").is_not_null()
        | pl.col("first_date_buy_age_group_0-3M").is_not_null()
        | pl.col("last_date_buy_milk4mom").is_not_null()
    )
)

pred_table
```

```python
# --- Tách 3 nhóm đối tượng tuổi/giai đoạn ---
group_step1 = pred_table.filter(pl.col("age_by_step1").is_not_null())
group_age03 = pred_table.filter(pl.col("age_by_age_group").is_not_null())
group_mom   = pred_table.filter(pl.col("age_by_milk4mom").is_not_null())

print("👶 Bé sơ sinh (Step 1):", group_step1.height, "khách hàng")
print("🍼 Bé 0–3 tháng:", group_age03.height, "khách hàng")
print("🤱 Mẹ bầu / sau sinh:", group_mom.height, "khách hàng")

# (tuỳ chọn) hiển thị vài dòng đầu mỗi nhóm
print("\n--- Bé sơ sinh (Step 1) ---")
display(group_step1.head(5))

print("\n--- Bé 0–3 tháng ---")
display(group_age03.head(5))

print("\n--- Mẹ bầu / sau sinh ---")
display(group_mom.head(5))
```

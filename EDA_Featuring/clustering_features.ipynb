```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

# ĐẶC TRƯNG 1

```python
# ==== 1) Chuẩn bị bảng giao dịch (mỗi lần mua) ====
ph = purchase_history
if "event_type" in ph.columns:
    ph = ph.filter(pl.col("event_type") == "Purchase")

txn = (
    ph.select(["customer_id", "price", "quantity", "created_date"])
      .with_columns([
          (pl.col("price").cast(pl.Float64) * pl.col("quantity").cast(pl.Float64).fill_null(1.0))
          .alias("transaction_value")
      ])
      .select(["customer_id", "created_date", "transaction_value"])
)

# ==== 2) Tính giá trị trung bình & phân cụm (từ code cũ, giữ nguyên logic) ====
feat = (
    txn.group_by("customer_id")
       .agg([
           pl.sum("transaction_value").alias("total_spending"),
           pl.max("transaction_value").alias("max_txn_value"),
           pl.mean("transaction_value").alias("transaction_value_avg"),
           pl.count("transaction_value").alias("purchase_count"),
       ])
)

# Phân cụm 3 nhóm
K = 3
pdf = feat.select(["total_spending", "transaction_value_avg"]).to_pandas()
scaler = StandardScaler()
X = scaler.fit_transform(pdf)

kmeans = KMeans(n_clusters=K, n_init=10, random_state=42)
labels = kmeans.fit_predict(X)

feat_pd = feat.to_pandas()
feat_pd["clustering_K"] = labels
feat = pl.from_pandas(feat_pd)

# ==== 3) Gán nhãn cụm theo mức chi tiêu ====
stats = (
    feat.group_by("clustering_K")
        .agg(pl.mean("total_spending").alias("mean_spending"))
        .sort("mean_spending")  # thấp -> cao
)
order = stats.select("clustering_K").to_series().to_list()
mapping_df = pl.DataFrame({
    "clustering_K": order,
    "segment": ["bình dân", "trung cấp", "cao cấp"],
})

feat = feat.join(mapping_df, on="clustering_K", how="left")

# ==== 4) Ghép từng giao dịch với đặc trưng trung bình & phân khúc ====
result_dactrung1 = (
    txn.join(feat.select(["customer_id", "transaction_value_avg", "segment"]), on="customer_id", how="left")
       .select(["customer_id", "created_date", "transaction_value", "transaction_value_avg", "segment"])
       .sort(["segment", "customer_id", "created_date"])
)

result_dactrung1.head(10)
```

```python
count_result = (result_dactrung1.group_by("segment").agg(pl.n_unique("customer_id").alias("n_customers")).sort("n_customers", descending=True))

count_result
```

# ĐẶC TRƯNG 2

```python
ph = purchase_history.select(["customer_id", "item_id", "timestamp"])


ph = ph.join(item.select(["item_id", "category_l1"]), on="item_id", how="left")


txn_category = (
    ph.group_by(["customer_id", "timestamp"])
      .agg(pl.col("category_l1").n_unique().alias("num_category"))
)

feat = (
    txn_category.group_by("customer_id")
                .agg(pl.mean("num_category").alias("avg_category_per_txn"))
)

pdf = feat.select(["avg_category_per_txn"]).to_pandas()

scaler = StandardScaler()
X = scaler.fit_transform(pdf)

kmeans = KMeans(n_clusters=3, n_init=10, random_state=42)
labels = kmeans.fit_predict(X)

feat_pd = feat.to_pandas()
feat_pd["clustering_K"] = labels
feat = pl.from_pandas(feat_pd)

stats = (
    feat.group_by("clustering_K")
        .agg(pl.mean("avg_category_per_txn").alias("mean_cat"))
        .sort("mean_cat")  
)
order = stats.select("clustering_K").to_series().to_list()
mapping_df = pl.DataFrame({
    "clustering_K": order,
    "segment": ["Mua ít", "Mua vừa", "Mua nhiều"],
})

feat = feat.join(mapping_df, on="clustering_K", how="left")
result_feature2 = (
    txn_category.join(feat.select(["customer_id", "avg_category_per_txn", "segment"]),
                      on="customer_id", how="left")
                 .select(["customer_id", "timestamp", "num_category", "avg_category_per_txn", "segment"])
                 .sort(["segment", "customer_id", "timestamp"])
)

result_feature2
```

```python
count_segment = result_feature2.group_by("segment").agg(pl.n_unique("customer_id").alias("n_customers"))
count_segment.sort("n_customers", descending=True)
```

# ĐẶC TRƯNG 3

```python


# ==== Bước 1: Gom nhóm giá mặt hàng sữa ====
# Lấy các item thuộc nhóm "sữa" (bao gồm cả "sua")
item_sua = (
    item.filter(pl.col("category_l1").str.to_lowercase().str.contains("sữa|sua"))
         .select(["item_id", "category_l1", "price"])
         .with_columns(pl.col("price").cast(pl.Float64))
         .drop_nulls(["item_id", "price"])
)

# Tính các ngưỡng tertile
q33, q66 = item_sua.select([
    pl.quantile("price", 0.33).alias("q33"),
    pl.quantile("price", 0.66).alias("q66")
]).row(0)

# Gán nhãn phân hạng giá
item_sua = (
    item_sua.with_columns(
        pl.when(pl.col("price") <= q33).then(pl.lit("Bình dân"))
         .when(pl.col("price") <= q66).then(pl.lit("Trung cấp"))
         .otherwise(pl.lit("Cao cấp"))
         .alias("price_segment")
    )
)

print(f"Ngưỡng giá SỮA: Bình dân ≤ {q33:.0f}, Trung cấp ≤ {q66:.0f}, Cao cấp > {q66:.0f}")

# ==== Bước 2: Gán nhãn khách hàng theo nhóm sữa mua nhiều nhất ====
# Join purchase_history với nhóm sữa
ph_sua = (
    purchase_history.join(item_sua.select(["item_id", "price_segment"]), on="item_id", how="inner")
)

# Gom nhóm theo khách hàng và nhãn giá
agg_sua = (
    ph_sua.group_by(["customer_id", "price_segment"])
           .agg(pl.col("quantity").cast(pl.Float64).sum().alias("total_qty"))
)

# Lấy nhãn có tổng quantity lớn nhất
segment_sua = (
    agg_sua.sort(["customer_id", "total_qty"], descending=[False, True])
            .unique(subset=["customer_id"], keep="first")
            .select(["customer_id", pl.col("price_segment").alias("segment_sua")])
)

segment_sua.head(10)
```

```python
count_sua = (
    segment_sua.group_by("segment_sua")
               .agg(pl.n_unique("customer_id").alias("n_customers"))
               .sort("n_customers", descending=True)
)
count_sua
```

# ĐẶC TRƯNG 4

```python
# ==== Bước 1: Gom nhóm giá mặt hàng tã ====
# Lấy các item thuộc nhóm "tã" (bao gồm cả "ta")
item_ta = (
    item.filter(pl.col("category_l1").str.to_lowercase().str.contains("tã|ta"))
         .select(["item_id", "category_l1", "price"])
         .with_columns(pl.col("price").cast(pl.Float64))
         .drop_nulls(["item_id", "price"])
)

# Tính ngưỡng tertile
q33_ta, q66_ta = item_ta.select([
    pl.quantile("price", 0.33).alias("q33"),
    pl.quantile("price", 0.66).alias("q66")
]).row(0)

# Gán nhãn phân hạng giá
item_ta = (
    item_ta.with_columns(
        pl.when(pl.col("price") <= q33_ta).then(pl.lit("Bình dân"))
         .when(pl.col("price") <= q66_ta).then(pl.lit("Trung cấp"))
         .otherwise(pl.lit("Cao cấp"))
         .alias("price_segment")
    )
)

print(f"Ngưỡng giá TÃ: Bình dân ≤ {q33_ta:.0f}, Trung cấp ≤ {q66_ta:.0f}, Cao cấp > {q66_ta:.0f}")

# ==== Bước 2: Gán nhãn khách hàng theo nhóm tã mua nhiều nhất ====
# Join purchase_history với nhóm tã
ph_ta = (
    purchase_history.join(item_ta.select(["item_id", "price_segment"]), on="item_id", how="inner")
)

# Gom nhóm theo khách hàng và nhãn giá
agg_ta = (
    ph_ta.group_by(["customer_id", "price_segment"])
          .agg(pl.col("quantity").cast(pl.Float64).sum().alias("total_qty"))
)

# Lấy nhãn có tổng quantity lớn nhất
segment_ta = (
    agg_ta.sort(["customer_id", "total_qty"], descending=[False, True])
           .unique(subset=["customer_id"], keep="first")
           .select(["customer_id", pl.col("price_segment").alias("segment_ta")])
)

segment_ta.head(10)
```

```python
count_ta = (
    segment_ta.group_by("segment_ta")
              .agg(pl.n_unique("customer_id").alias("n_customers"))
              .sort("n_customers", descending=True)
)
count_ta
```
